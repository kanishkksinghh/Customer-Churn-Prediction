```python
df["Onboard_date"] = pd.to_datetime(
    df["Onboard_date"]
)
```

```python
df["Onboard_date"] = df["Onboard_date"].dt.year
```

```python
df["Location"].value_counts()
```

```python
location_churn = pd.crosstab(
    df["Location"],
    df["Churn"],
    normalize="index"
) * 100

print(location_churn)
```

```python
df["Purchase_Per_Year"] = ( df["Total_Purchase"] / df["Years"].replace(0, np.nan)
)
```

```python
df["Sites_Usage_Level"] = pd.cut(
    df["Num_Sites"], 
    bins=[0, 5, 10, float("inf")], 
    labels=["low", "Medium", "High"]
)
```

```python
features = [
    "Age",
    "Total_Purchase",
    "Account_Manager",
    "Years",
    "Num_Sites"
]

X = df[features]
y = df["Churn"]
```

```python
print("X shape:", X.shape)
print("y shape:", y.shape)
```

```python
print("\nFeatues: ")
print(X.head())
```

```python
print("\nTarget: ")
print(y.head())
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, 
    y,
    test_size=0.20, random_state=42, stratify=y
)
```

```python
print("X_train:", X_train.shape)
print("X_test:", X_test.shape)

print("y_train:", y_train.shape)
print("y_test:", y_test.shape)
```

```python
print("Training Churn rate:")
print(y_train.mean())

print("Testing Churn rate:")
print(y_test.mean())
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()
```

```python
X_train_scaled = scaler.fit_transform(X_train)
```

```python
X_test_scaled = scaler.fit_transform(X_test)
```

```python
print(X_train_scaled[:5]) #standardized value
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
model = LogisticRegression(random_state=42)
```

```python
model.fit(X_train_scaled, y_train)
```

```python
y_predict = model.predict(X_test_scaled)
```

```python
print(y_predict[:20])
```

```python
y_test
```

```python
y_probability = model.predict_proba(X_train_scaled)
```

```python
print(y_probability)
```

```python
churn_probabilty = model.predict_proba(X_train_scaled) [:, 1]

print(churn_probabilty[:10])
```
